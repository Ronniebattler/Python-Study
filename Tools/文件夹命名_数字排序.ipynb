# 数字排序命名

```python
import os

def rename_pdfs(directory):
    # 获取指定目录下所有的PDF文件
    files = [f for f in os.listdir(directory) if f.endswith('.pdf')]
    files.sort()  # 可以根据需要排序

    # 为每个PDF文件添加编号
    for i, file in enumerate(files, start=1):
        new_filename = f"{i}_{file}"  # 直接使用数字序号
        os.rename(os.path.join(directory, file), os.path.join(directory, new_filename))

# 使用示例：
# 假设你的PDF文件在'path/to/your/folder'文件夹中
rename_pdfs(r'D:\WPS_Renhe\研究生\2024社科\金融风险\文献\文献\汇总')
```

# 删除文件名前面的数字和下划线

```python
import os
import re

def rename_pdf_files(folder_path):
    # 遍历文件夹中的所有文件
    for filename in os.listdir(folder_path):
        if filename.endswith('.pdf'):  # 确保文件是PDF文件
            old_filepath = os.path.join(folder_path, filename)
            new_filename = re.sub(r'^\d+_', '', filename)  # 删除开头的数字和下划线
            new_filepath = os.path.join(folder_path, new_filename)
            os.rename(old_filepath, new_filepath)  # 重命名文件

folder_path = r'D:\WPS_Renhe\研究生\2024社科\金融风险\文献\文献\汇总'  # 替换成你的文件夹路径
rename_pdf_files(folder_path)
```
